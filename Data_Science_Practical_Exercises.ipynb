```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Asha", "Ravi", "Meena", "Asha", "Arjun"],
    "Department": ["CSM", "CSE", "CSM", "CSM", "ECE"],
    "Age": [19, 20, np.nan, 19, 21],
    "Marks": [82, 71, 91, 82, np.nan]
}

df_dirty = pd.DataFrame(data)

display(df_dirty)
```

```python
df_dirty.isnull().sum()
```

```python
df_dirty.duplicated()
```

```python
df_dirty.dtypes
```

```python

df_clean = df_dirty.drop_duplicates().copy()

df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Marks"] = df_clean["Marks"].fillna(df_clean["Marks"].median())

print(display(df_clean))

print(df_clean.isnull().sum())
```

```python
print("Missing Age values:", df_dirty["Age"].isnull().sum())
print("Missing Marks values:", df_dirty["Marks"].isnull().sum())
print("Duplicate rows:", df_dirty.duplicated().sum())
print("Rows after removing duplicates:", len(df_dirty.drop_duplicates()))
```

```python
print("Average Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Standard Deviation:", df["Marks"].std())
```

```python
department_average = df.groupby("Department")["Marks"].mean()

print(department_average)
```

```python
department_summary = df.groupby("Department").agg(
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Lowest_Marks=("Marks", "min"),
    Average_Study_Hours=("Study_Hours", "mean")
)

display(department_summary)
```

```python
highest_student = df.loc[df["Marks"].idxmax()]

print("Highest Scoring Student:")
display(highest_student)

lowest_student = df.loc[df["Marks"].idxmin()]

print("Lowest Scoring Student:")
display(lowest_student)
```

```python
students_need_improvement = df[df["Marks"] < 70]

display(students_need_improvement)
```

```python
df["Performance"] = np.where(
    df["Marks"] >= 80,
    "Excellent",
    np.where(
        df["Marks"] >= 60,
        "Good",
        "Needs Improvement"
    )
)

display(df)
```

```python
print("===== DATA ANALYSIS SUMMARY =====")

print("Total students:", len(df))
print("Average marks:", round(df["Marks"].mean(), 2))
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())

print("\nDepartment-wise average marks:")
display(
    df.groupby("Department")["Marks"]
      .mean()
      .round(2)
      .to_frame("Average Marks")
)

print("\nStudy Hours vs Marks correlation:",
      round(df["Study_Hours"].corr(df["Marks"]), 2))
```
