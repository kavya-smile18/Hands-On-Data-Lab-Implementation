```python

import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Meena", "Arjun", "Divya"],
    "Department": ["CSM", "CSE", "CSM", "ECE", "CSE"],
    "Study_Hours": [3, 2, 4, 1, 5],
    "Marks": [82, 71, 91, 59, 95]
}

df = pd.DataFrame(data)

display(df)
```

```python
print(df.head())
print("shape =" ,df.shape)

print(df.describe())

# Select students who scored more than 80
df[df["Marks"] > 80]

# Sort students by marks, highest first
df.sort_values("Marks", ascending=False)
```

```python
csm_students = df[df["Department"] == "CSM"]
print(csm_students)
```

```python
hoursstudy = df[df["Study_Hours"] >= 3]
print(hoursstudy)
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Asha", "Ravi", "Meena", "Asha", "Arjun"],
    "Department": ["CSM", "CSE", "CSM", "CSM", "ECE"],
    "Age": [19, 20, np.nan, 19, 21],
    "Marks": [82, 71, 91, 82, np.nan]
}

df_dirty = pd.DataFrame(data)

display(df_dirty)
```

```python
df_dirty.isnull().sum()
```

```python
df_dirty.duplicated()
```

```python
df_dirty.dtypes
```

```python

df_clean = df_dirty.drop_duplicates().copy()

df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Marks"] = df_clean["Marks"].fillna(df_clean["Marks"].median())

print(display(df_clean))

print(df_clean.isnull().sum())
```

```python
print("Missing Age values:", df_dirty["Age"].isnull().sum())
print("Missing Marks values:", df_dirty["Marks"].isnull().sum())
print("Duplicate rows:", df_dirty.duplicated().sum())
print("Rows after removing duplicates:", len(df_dirty.drop_duplicates()))
```

```python
import matplotlib.pyplot as plt

names = ["Asha", "Ravi", "Meena", "Arjun", "Divya"]
marks = [82, 71, 91, 59, 95]

plt.bar(names, marks)
plt.title("Student Marks Comparison")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()
```

```python
plt.hist(marks, bins=5, edgecolor="black")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
study_hours = [3, 2, 4, 1, 5]

plt.scatter(study_hours, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```
